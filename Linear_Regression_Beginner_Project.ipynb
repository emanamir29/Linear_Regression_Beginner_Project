```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sb
from sklearn.linear_model import LinearRegression as lg
```

```python
ali= pd.DataFrame(np.random.rand(6,2),
                  np.arange(6),
                  columns=['eman','zara'])
ali
```

```python
x=ali.eman
y=ali.zara
```

```python
plt.figure(figsize=(5,3))
plt.xlabel('eman')
plt.ylabel('zara')
plt.scatter(x,y,color='g',marker='o')
plt.show()
```

```python
ali.corr()
ali
```

```python
X=ali.drop('eman',axis='columns')
X
```

```python
anna= lg()
anna.fit(X.values,y.values)
```

```python
anna.score(X.values,y.values)
```
